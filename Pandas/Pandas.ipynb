```python
#With Python

import csv

def calculate_stats(data, industry=None):
    ratings = []

    for row in data:
        if row[3].upper() != "NULL" and (not industry or row[2] == industry):
            ratings.append(float(row[3]))

    max_rating = max(ratings)
    min_rating = min(ratings)
    average = sum(ratings) / len(ratings)

    return max_rating, min_rating, average


with open ('movies.csv') as f:
    data= list(csv.reader(f))
    header= data[0]
    data= data[1:]
    print(header)
    print(data)

max_rating, min_rating, average= calculate_stats(data)

print(f"Max_Ratings= {max_rating}, Min_Rating= {min_rating}, Average= {average}")

```

```python
#Now look at currency column there different types of $$, Dollars like that we need to convert them into " USD "

#first create a function that converts then later call that function while reading the file

def standarize_currency(curr):
  if curr == "$$" or curr == "Dollars":
    return "USD"
  return curr


#Now reading the file

df_finacials1= pd.read_excel("movies_db.xlsx", "financials", converters= {"currency": standarize_currency})
df_finacials1.head(4)
```

```python
#merging to shhets into one data frame

df_mergedata= pd.merge(df_movies, df_finacials1, on= 'movie_id' )
df_mergedata.head(5)
```
