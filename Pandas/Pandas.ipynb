```python
#With Python

import csv

def calculate_stats(data, industry=None):
    ratings = []

    for row in data:
        if row[3].upper() != "NULL" and (not industry or row[2] == industry):
            ratings.append(float(row[3]))

    max_rating = max(ratings)
    min_rating = min(ratings)
    average = sum(ratings) / len(ratings)

    return max_rating, min_rating, average


with open ('movies.csv') as f:
    data= list(csv.reader(f))
    header= data[0]
    data= data[1:]
    print(header)
    print(data)

max_rating, min_rating, average= calculate_stats(data)

print(f"Max_Ratings= {max_rating}, Min_Rating= {min_rating}, Average= {average}")

```

```python
#Now look at currency column there different types of $$, Dollars like that we need to convert them into " USD "

#first create a function that converts then later call that function while reading the file

def standarize_currency(curr):
  if curr == "$$" or curr == "Dollars":
    return "USD"
  return curr


#Now reading the file

df_finacials1= pd.read_excel("movies_db.xlsx", "financials", converters= {"currency": standarize_currency})
df_finacials1.head(4)
```

```python
#merging to shhets into one data frame

df_mergedata= pd.merge(df_movies, df_finacials1, on= 'movie_id' )
df_mergedata.head(5)
```

```python
#convert this data frame to excel file and naming sheet to sheet1

df_mergedata.to_excel("movies_merged.xlsx", sheet_name="sheet1", index=False)
```

```python
#replacing NaN values with mean

df_weathermean= df_weather1.fillna({
    "temperature": df_weather1.temperature.mean(),
    "windspeed": df_weather1.windspeed.mean(),
    "event": "No Event"
})

df_weathermean
```

```python
#Now Median

df_weathermedian= df_weather1.fillna({
    "temperature": df_weather1.temperature.median(),
    "windspeed": df_weather1.windspeed.median(),
    "event": "No Event"
})

df_weathermedian
```

```python
#ffill ----> forward fill

# df_weatherffill= df_weather1.fillna(method="ffill")
# df_weatherffill

df_weatherffill= df_weather1.ffill()
df_weatherffill
```

```python
#bfill ---> Backward fill

df_weatherbfill= df_weather1.bfill()
df_weatherbfill
```

```python
#by column

df_weatherbfill= df_weather1.bfill(axis="columns")
df_weatherbfill
```

```python
#interpolate() is one of the smartest ways to fill missing values.
#Instead of copying the previous value (ffill) or the next value (bfill), it estimates the missing value based on the surrounding values.

df_weatherinterpolate= df_weather1.interpolate()
df_weatherinterpolate
```

```python
#dropna values row based droping

df_weatherdropna= df_weather1.dropna()
df_weatherdropna
```
